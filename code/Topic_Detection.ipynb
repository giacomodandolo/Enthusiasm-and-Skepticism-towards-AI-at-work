```python
from SemanticNetwork import SemanticNetwork
from util import export_igraph_to_csv, save_topic_word_csv, save_C_to_csv, show_wordcloud, extract_metrics
import pandas as pd
import numpy as np
import os
os.environ["TOKENIZERS_PARALLELISM"] = "false"
```

```python
def check_metrics(df_metrics): 
    return ( 
        (df_metrics["NMI"] > 0.15) & 
        (df_metrics["Q"] > 0.4) & 
        (df_metrics["Ncut"] < 0.2) & 
        (df_metrics["Infomap"] > 0.5) & 
        (df_metrics["com"] > 0.5)
    )

def metric_score(df_metrics): 
    m = df_metrics 
    score = ( m["NMI"] + m["Q"] + m["Infomap"] + m["com"] - m["Ncut"] ) 
    return score
```

```python
files = ["programmers", "designers"]
for file in files:
    # Read file data
    df = pd.read_csv(f"./data/{file}.csv")
    df.drop_duplicates(subset=["translated_clean"],inplace=True)
    df.reset_index(drop=True, inplace=True)

    # Create semantic network
    sn = SemanticNetwork()
    sn.clean_text(df, content_column="translated_clean", kr_list=["ADJ","NOUN","VERB"], to_keep=True)
    sn.get_Nwd(n_min=10)
    # Create words graph
    sn.build_word_graph(threshold=1e-8)
    # Create documents graph
    sn.build_document_graph(threshold=1e-8)
    # Export graph
    export_igraph_to_csv(sn.G_word, f"./output/{file}/word_nodes.csv", f"./output/{file}/word_edges.csv")

    # Condition parameters
    acceptable = False
    attempt = 0
    max_attempts = 10

    best_score = -np.inf
    best_louvain = None
    best_louvain_overlapping = None
    best_BERTopic = None

    while not acceptable and attempt < max_attempts:
        attempt += 1
        print(f"Attempt {attempt} for {file}:")
        # Run Louvain
        print("RUN LUOVAIN")
        sn.detect_communities()
        # Run Louvain (overlapping)
        # print("RUN LOUVAIN (OVERLAPPING)")
        # sn.detect_communities_overlapping()
        # Run BERTopic
        print("RUN BERTOPIC")
        sn.run_bertopic(content_column="clean list")

        # Test Louvain metrics
        print("COMPUTE METRICS LOUVAIN")
        louvain_metrics = extract_metrics(
            sn.louvain_Pwc,
            sn.louvain_Pcc,
            sn.louvain_C,
            sn.Pdd,
            sn.louvain_pc,
            f"./output/{file}/louvain/metrics.csv",
            "louvain"
        )
        print(louvain_metrics)
        louvain_score = metric_score(louvain_metrics.iloc[0])
        print(f"Score: {louvain_score}")
        if louvain_score > best_score:
            best_score = louvain_score
            best_louvain = {
                "Pwc": sn.louvain_Pwc.copy(),
                "Pcc": sn.louvain_Pcc.copy(),
                "C": sn.louvain_C.copy(),
                "metrics": louvain_metrics.copy()
            }
        louvain_acceptable = check_metrics(louvain_metrics.iloc[0])
        print(f"Acceptable: {louvain_acceptable}")

        # Test Louvain (overlapping) metrics
        """
        print("COMPUTE METRICS LOUVAIN (OVERLAPPING)")
        louvain_overlapping_metrics = extract_metrics(
            sn.louvain_Pwc_overlapping,
            sn.louvain_Pcc_overlapping,
            sn.louvain_C_overlapping,
            sn.Pdd,
            sn.louvain_pc_overlapping,
            f"./output/{file}/louvain_overlapping/metrics.csv",
            "louvain_overlapping"
        )
        print(louvain_overlapping_metrics)
        louvain_overlapping_score = metric_score(louvain_overlapping_metrics.iloc[0])
        print(f"Score: {louvain_overlapping_score}")
        if louvain_overlapping_score > best_score:
            best_score = louvain_overlapping_score
            best_louvain_overlapping = {
                "Pwc": sn.louvain_Pwc_overlapping.copy(),
                "Pcc": sn.louvain_Pcc_overlapping.copy(),
                "C": sn.louvain_C_overlapping.copy(),
                "metrics": louvain_overlapping_metrics.copy()
            }
        louvain_overlapping_acceptable = check_metrics(louvain_overlapping_metrics.iloc[0])
        print(f"Acceptable: {louvain_overlapping_acceptable}")
        """
        # Test BERTopic metrics
        print("COMPUTE METRICS BERTOPIC")
        BERTopic_metrics = extract_metrics(sn.BERTopic_Pwc, sn.BERTopic_Pcc, sn.BERTopic_C, sn.Pdd, sn.BERTopic_pc, f"./output/{file}/BERTopic/metrics.csv", "BERTopic")
        print(BERTopic_metrics)
        BERTopic_score = metric_score(BERTopic_metrics.iloc[0])
        print(f"Score: {BERTopic_score}")
        if BERTopic_score > best_score:
            best_score = BERTopic_score
            best_BERTopic = {
                "Pwc": sn.BERTopic_Pwc.copy(),
                "Pcc": sn.BERTopic_Pcc.copy(),
                "C": sn.BERTopic_C.copy(),
                "metrics": BERTopic_metrics.copy()
            }
        BERTopic_acceptable = check_metrics(BERTopic_metrics.iloc[0])
        print(f"Acceptable: {BERTopic_acceptable}")
        acceptable = louvain_acceptable or BERTopic_acceptable

    if not acceptable:
        print("Max attempts reached. Saving best run:")
        if best_louvain is not None:
            sn.louvain_Pwc = best_louvain["Pwc"]
            sn.louvain_Pcc = best_louvain["Pcc"]
            sn.louvain_C = best_louvain["C"]
            best_louvain["metrics"].to_csv(f"./output/{file}/louvain/metrics_best.csv", index=False)
        """
        if best_louvain_overlapping is not None:
            sn.louvain_Pwc_overlapping = best_louvain_overlapping["Pwc"]
            sn.louvain_Pcc_overlapping = best_louvain_overlapping["Pcc"]
            sn.louvain_C_overlapping = best_louvain_overlapping["C"]
            best_louvain_overlapping["metrics"].to_csv(f"./output/{file}/louvain_overlapping/metrics_best.csv")
        """
        if best_BERTopic is not None:
            sn.BERTopic_Pwc = best_BERTopic["Pwc"]
            sn.BERTopic_Pcc = best_BERTopic["Pcc"]
            sn.BERTopic_C = best_BERTopic["C"]
            best_BERTopic["metrics"].to_csv(f"./output/{file}/BERTopic/metrics_best.csv", index=False)

    print("SAVE LOUVAIN WORDCLOUD")
    show_wordcloud(sn.louvain_Pwc, sn.words, sn.word_pagerank, f"./output/{file}/louvain/louvain_wordcloud.png")
    # print("SAVE LOUVAIN OVERLAPPING WORDCLOUD")
    # show_wordcloud(sn.louvain_Pwc_overlapping, sn.words, sn.word_pagerank, f"./output/{file}/louvain_overlapping/louvain_overlapping_wordcloud.png")
    print("SAVE BERTOPIC WORDCLOUD")
    show_wordcloud(sn.BERTopic_Pwc, sn.words, sn.word_pagerank, f"./output/{file}/BERTopic/BERTopic_wordcloud.png")
    print("SAVE TOPIC LOUVAIN")
    save_topic_word_csv(sn.louvain_Pwc, sn.words, sn.word_pagerank, f"./output/{file}/louvain", top_k=100)
    # print("SAVE TOPIC LOUVAIN OVERLAPPING")
    # save_topic_word_csv(sn.louvain_Pwc_overlapping, sn.words, sn.word_pagerank, f"./output/{file}/louvain_overlapping")
    print("SAVE TOPIC BERTOPIC")
    save_topic_word_csv(sn.BERTopic_Pwc, sn.words, sn.word_pagerank, f"./output/{file}/BERTopic/", top_k=100)
    print("SAVE DOCUMENTS TO TOPIC MATRIX LOUVAIN")
    save_C_to_csv(sn.louvain_C, f"./output/{file}/louvain/C.csv", document_names=sn.documents)
    # print("SAVE DOCUMENTS TO TOPIC MATRIX LOUVAIN OVERLAPPING")
    # save_C_to_csv(sn.louvain_C_overlapping, f"./output/{file}/louvain_overlapping/C.csv", document_names=sn.documents)
    print("SAVE DOCUMENTS TO TOPIC MATRIX BERTOPIC")
    save_C_to_csv(sn.BERTopic_C, f"./output/{file}/BERTopic/C.csv", document_names=sn.documents)
```
